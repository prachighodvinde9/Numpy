```python
# 41 Generate marks between 0 and 100 for 50 students

marks = np.random.randint(0, 101, 50)

print(marks)
```

```python
# Calculate statistics for marks array
#42 calculate total,average, highest,lowest.
print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

```python
#Count Above 75, Below 35, Between 35–75
# Count students scoring above 75
above_75 = np.sum(marks > 75)

# Count students scoring below 35
below_35 = np.sum(marks < 35)

# Count students scoring between 35 and 75
between_35_75 = np.sum((marks >= 35) & (marks <= 75))

print("Above 75:", above_75)
print("Below 35:", below_35)
print("Between 35 and 75:", between_35_75)
```

```python
# 44 Assign grades based on marks

grades = np.where(marks >= 75, "A",
         np.where(marks >= 60, "B",
         np.where(marks >= 35, "C", "F")))

print(grades)
```

```python
# 45 display top 5 marks

top_5 = np.sort(marks)[-5:]

print(top_5)
```

```python
#46 Display bottom 5 marks

bottom_5 = np.sort(marks)[:5]

print(bottom_5)
```

```python
#47 Find median

print("Median:", np.median(marks))

#  standard deviation

print("Standard Deviation:", np.std(marks))
```

```python
# 48 Create 5-Subject Marks Matrix and Calculate Totals
#  Generate marks for 10 students in 5 subjects

subject_marks = np.random.randint(35, 101, (10, 5))

# Calculate total marks of each student

totals = np.sum(subject_marks, axis=1)

print("Marks Matrix:")
print(subject_marks)

print("\nTotal Marks:")
print(totals)
```

```python
# 49 Find Topper in Each Subject

subject_toppers = np.max(subject_marks, axis=0)

print(subject_toppers)
```

```python
# 50 Generate Report Card with Total, Average, Percentage and Grade
# Calculate average and percentage

average = np.mean(subject_marks, axis=1)
percentage = average

# Assign grades

grades = np.where(percentage >= 75, "A",
         np.where(percentage >= 60, "B",
         np.where(percentage >= 35, "C", "F")))

# Display report card

for i in range(len(subject_marks)):
    print("Student", i + 1)
    print("Total:", totals[i])
    print("Average:", round(average[i], 2))
    print("Percentage:", round(percentage[i], 2))
    print("Grade:", grades[i])
    print("-" * 20)
```
