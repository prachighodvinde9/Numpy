```python
# 1 Create a file and write text into it
file = open("hello.txt", "w")
file.write("Hello Python")
file.close()

print("Data written successfully.")
```

```python
# Open file in read mode
file = open("hello.txt", "r")

# 2 Read and display line by line
for line in file:
    print(line.strip())

file.close()
```

```python
# 3. Count Lines, Words, and Characters in text file.
# Open file
file = open("hello.txt", "r")

data = file.read()

lines = data.split("\n")
words = data.split()

print("Lines:", len(lines))
print("Words:", len(words))
print("Characters:", len(data))

file.close()
```

```python
# 4 Append Student Names to an Existing File
# Open file in append mode
file = open("students.txt", "a")

students = ["Prachi", "Riya", "Amit", "Sneha"]

for student in students:
    file.write(student + "\n")

file.close()

print("Names appended successfully.")
```

```python
#5. Copy Contents of One File to Another
# Open source file
source = open("hello.txt", "r")

# Read contents
data = source.read()

# Open destination file
destination = open("copy.txt", "w")

# Write contents
destination.write(data)

source.close()
destination.close()

print("File copied successfully.")

# Open source file
source = open("hello.txt", "r")

# Read contents
data = source.read()

# Open destination file
destination = open("copy.txt", "w")

# Write contents
destination.write(data)

source.close()
destination.close()

print("File copied successfully.")
```

```python
# 6. Write Numbers 1 to 100 into a File
# Open file in write mode
file = open("numbers.txt", "w")

for i in range(1, 101):
    file.write(str(i) + "\n")

file.close()

print("Numbers written successfully.")
```

```python
# 7. Search for a Word and Print Line Number
# Word to search
word = input("Enter word to search: ")

file = open("hello.txt", "r")

for line_no, line in enumerate(file, start=1):
    if word in line:
        print("Found at line:", line_no)

file.close()

# Word to search
word = input("Enter word to search: ")

file = open("hello.txt", "r")

for line_no, line in enumerate(file, start=1):
    if word in line:
        print("Found at line:", line_no)

file.close()
```

```python
#8. Replace a Word and Save as New File
# Open original file
file = open("hello.txt", "r")

data = file.read()

# Replace word
new_data = data.replace("Python", "Programming")

# Save in new file
new_file = open("newfile.txt", "w")
new_file.write(new_data)

file.close()
new_file.close()

print("Word replaced successfully.")
```

```python
# 9 Display Lines Containing More Than 3 Words
file = open("hello.txt", "w")

file.write("Hello Python Programming Language\n")
file.write("Good Morning Everyone\n")
file.write("Welcome\n")

file.close()
```

```python
# Open file in read mode
file = open("hello.txt", "r")

# Read file line by line
for line in file:

    # Check if line contains more than 3 words
    if len(line.split()) > 3:
        print(line.strip())

# Close file
file.close()
```

```python
# 10. Store User Input Until "exit" is Entered
# Open file in write mode
file = open("userdata.txt", "w")

while True:
    text = input("Enter text: ")

    if text.lower() == "exit":
        break

    file.write(text + "\n")

file.close()

print("Data saved successfully.")
```
