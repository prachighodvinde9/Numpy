```python
import numpy as np
```

```python
# 1 Generate marks between 0 and 100 for 50 students
marks = np.random.randint(0, 101, 50)

print(marks)
```

```python
#42. Calculate Total, Average, Highest, Lowest
# Total Marks
print("Total Marks:", np.sum(marks))

# Average Marks
print("Average Marks:", np.mean(marks))

# Highest Marks
print("Highest Marks:", np.max(marks))

# Lowest Marks
print("Lowest Marks:", np.min(marks))
```

```python
# 43. Count Above 75, Below 35, Between 35–75
# Count students above 75
print("Above 75:", np.sum(marks > 75))

# Count students below 35
print("Below 35:", np.sum(marks < 35))

# Count students between 35 and 75
print("Between 35 and 75:", np.sum((marks >= 35) & (marks <= 75)))
```

```python
# 44.Assign Grades
grades = np.where(marks >= 75, "A",
         np.where(marks >= 60, "B",
         np.where(marks >= 35, "C", "F")))

print(grades)
```

```python
#45 Display Top 5 Marks
top5 = np.sort(marks)[-5:]

print("Top 5 Marks:", top5)
```

```python
# 46. Display Bottom 5 Marks
bottom5 = np.sort(marks)[:5]

print("Bottom 5 Marks:", bottom5)
```

```python
#47. Find Median and Standard Deviation
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
```

```python
#48. Create 5-Subject Marks Matrix and Calculate Totals
# Marks of 10 students in 5 subjects
subject_marks = np.random.randint(35, 101, (10, 5))

print("Marks Matrix:")
print(subject_marks)

# Total marks of each student
totals = np.sum(subject_marks, axis=1)

print("Student Totals:")
print(totals)
```

```python
#49. Find Topper in Each Subject
# Highest marks in each subject
topper_subject = np.max(subject_marks, axis=0)

print("Topper Marks in Each Subject:")
print(topper_subject)
```

```python
#50. Generate Report Card with Total, Average, Percentage and Grade
# Average and Percentage
average = np.mean(subject_marks, axis=1)
percentage = average

# Grade Calculation
grades = np.where(percentage >= 75, "A",
         np.where(percentage >= 60, "B",
         np.where(percentage >= 35, "C", "F")))

# Report Card
for i in range(len(subject_marks)):
    print("\nStudent", i + 1)
    print("Total Marks:", totals[i])
    print("Average:", round(average[i], 2))
    print("Percentage:", round(percentage[i], 2))
    print("Grade:", grades[i])
```
